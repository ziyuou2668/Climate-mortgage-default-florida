## Setup

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("your/data/path/here")

OUT_HUD_ZIP3  = DATA_DIR / "zip3_county_weighted.parquet"
OUT_FEMA_FL   = DATA_DIR / "fema_fl_hurricanes_county_year.parquet"
OUT_FANNIE_FL = DATA_DIR / "fannie_mae_2017Q1_FL.parquet"
```

## Load and summarise each output

```python
for name, path in [
    ('HUD ZIP3-county weighted mapping', OUT_HUD_ZIP3),
    ('FEMA FL hurricane county-year',    OUT_FEMA_FL),
    ('Fannie Mae FL 2017Q1 panel',       OUT_FANNIE_FL),
]:
    if not path.exists():
        print(f"[MISSING] {name}: {path}")
        print()
        continue
    df = pd.read_parquet(path)
    size_mb = path.stat().st_size / 1e6
    print(f"{name}")
    print(f"  Path:    {path}")
    print(f"  Rows:    {len(df):,}")
    print(f"  Columns: {len(df.columns)}")
    print(f"  Size:    {size_mb:.1f} MB")
    print(f"  Sample columns: {list(df.columns[:8])}")
    print()
```
